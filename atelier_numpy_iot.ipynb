### Partie 1 : Listes Python vs tableaux NumPy 

On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8 

#### 1) Créer une liste Python stockant ces températures

```python
tprt = [24.5, 25.1, 26.3, 27.0, 25.8]
```

#### 2) Création d'un tableau NumPy équivalent stockant ces températures

```python
# Importation de la bibliothèque NumPy
import numpy as np
```

```python
tprt_array = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
```

#### 3) Multiplication par 2 sur ces deux collections

```python
# Multiplication de chaque élément de la liste par 2
mul_tprt_list = [temp * 2 for temp in tprt]
mul_tprt_list
```

```python
# Multiplication de chaque élément du tableau NumPy par 2
mul_tprt_array = tprt_array * 2
mul_tprt_array
```

#### 4) Différence observée ?

On remarque que le liste python 'tprt' utilise une boucle 'for' en multipliant chaque élément de la liste pour pouvoir
avoir le résulat de la multiplication par 2 des températures stockés. Par contre le vecteur Numpy 'tprt_array' multiplie directement et simultanément tous éléments par 2 (technique du 'broadcasting') sans passer par une boucle. Ce qui énorme un terme de gain de temps
